```python

import pandas as pd

# Charger le fichier CSV
data = pd.read_csv(
    r"C:\Users\pc\Documents\Scrapped_sentimental_data.csv"
)

# Garder uniquement les sentiments positifs et négatifs
data_binary = data[
    data["sentiment_category"].isin(["positive", "negative"])
].copy()

# Supprimer les lignes sans texte ou sans sentiment
data_binary = data_binary.dropna(
    subset=["text", "sentiment_category"]
)

print("Nombre total de textes :", len(data_binary))
print(data_binary["sentiment_category"].value_counts())
```

```python

import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def nettoyer_texte(texte):
    texte = texte.lower()
    texte = re.sub(r"http\S+|www\S+", "", texte)
    texte = re.sub(r"[^a-z\s]", " ", texte)
    mots = word_tokenize(texte)

    mots = [
        lemmatizer.lemmatize(mot)
        for mot in mots
        if mot not in stop_words and len(mot) > 1
    ]

    return " ".join(mots)

data_binary["text_clean"] = data_binary["text"].apply(nettoyer_texte)

print(data_binary[["text", "text_clean"]].head())
print("Nombre total de textes :", len(data_binary))
```

```python

from sklearn.model_selection import train_test_split

X = data_binary["text_clean"]
y = data_binary["sentiment_category"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Textes d'entraînement :", len(X_train))
print("Textes de test :", len(X_test))
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Créer le vectoriseur TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)

# Apprendre le vocabulaire et transformer les textes d'entraînement
X_train_tfidf = vectorizer.fit_transform(X_train)

# Transformer les textes de test avec le même vocabulaire
X_test_tfidf = vectorizer.transform(X_test)

# Afficher les dimensions
print("Dimension entraînement :", X_train_tfidf.shape)
print("Dimension test :", X_test_tfidf.shape)
print("Nombre de caractéristiques :", len(vectorizer.get_feature_names_out()))
```

```python

# Importer les algorithmes
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

# 1. Créer les modèles
modele_lr = LogisticRegression(max_iter=1000)
modele_nb = MultinomialNB()
modele_svm = LinearSVC()

# 2. Entraîner les modèles
modele_lr.fit(X_train_tfidf, y_train)
modele_nb.fit(X_train_tfidf, y_train)
modele_svm.fit(X_train_tfidf, y_train)

print("Les 3 modèles ont été entraînés avec succès !")
```

```python

import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Liste des modèles entraînés
modeles = {
    "Logistic Regression": modele_lr,
    "Multinomial Naive Bayes": modele_nb,
    "Linear SVM": modele_svm
}

# Préparer la liste des résultats
resultats = []

# Évaluer chaque modèle
for nom, modele in modeles.items():

    # Prédire les sentiments des textes de test
    y_pred = modele.predict(X_test_tfidf)

    # Calculer les métriques
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, pos_label="positive"
    )
    recall = recall_score(
        y_test, y_pred, pos_label="positive"
    )
    f1 = f1_score(
        y_test, y_pred, pos_label="positive"
    )

    # Enregistrer les résultats
    resultats.append({
        "Algorithme": nom,
        "Accuracy": round(accuracy, 4),
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1-score": round(f1, 4)
    })

    # Afficher le rapport détaillé
    print("\n" + "=" * 50)
    print("Algorithme :", nom)
    print(classification_report(y_test, y_pred))

# Afficher le tableau comparatif
tableau_resultats = pd.DataFrame(resultats)

print("\nTABLEAU COMPARATIF")
display(tableau_resultats)
```

```python

import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

# Définir les 3 modèles
modeles_cv = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Multinomial Naive Bayes": MultinomialNB(),
    "Linear SVM": LinearSVC()
}

# Validation croisée à 5 plis
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

resultats_cv = []

for nom, modele in modeles_cv.items():

    # Pipeline : TF-IDF + algorithme
    pipeline = Pipeline([
        ("tfidf", TfidfVectorizer(max_features=5000)),
        ("modele", modele)
    ])

    # Calculer les métriques sur les 5 plis
    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "accuracy": "accuracy",
            "precision": "precision_macro",
            "recall": "recall_macro",
            "f1": "f1_macro"
        },
        n_jobs=-1
    )

    # Calculer les moyennes
    resultats_cv.append({
        "Algorithme": nom,
        "Accuracy moyenne": round(scores["test_accuracy"].mean(), 4),
        "Precision moyenne": round(scores["test_precision"].mean(), 4),
        "Recall moyen": round(scores["test_recall"].mean(), 4),
        "F1-score moyen": round(scores["test_f1"].mean(), 4)
    })

# Afficher le tableau final
tableau_cv = pd.DataFrame(resultats_cv)
display(tableau_cv)
```
